```python
!pip install -q ultralytics

import os, glob, shutil, random
import xml.etree.ElementTree as ET

src_xml = "/content/dataset/annotations"
out = "/content/pothole_yolo"

# görüntüleri otomatik bul
imgs = {}
for p in glob.glob("/content/dataset/**/*", recursive=True):
    if p.lower().endswith((".jpg", ".jpeg", ".png")):
        imgs[os.path.basename(p)] = p
print("görüntü sayısı:", len(imgs))

files = sorted(glob.glob(f"{src_xml}/*.xml"))
random.seed(42); random.shuffle(files)
n = len(files)
splits = {"train": files[:int(.7*n)], "val": files[int(.7*n):int(.85*n)], "test": files[int(.85*n):]}

skipped = 0
for split, items in splits.items():
    os.makedirs(f"{out}/images/{split}", exist_ok=True)
    os.makedirs(f"{out}/labels/{split}", exist_ok=True)
    for xf in items:
        r = ET.parse(xf).getroot()
        name = r.find("filename").text
        if name not in imgs:
            alt = os.path.splitext(os.path.basename(xf))[0]
            name = next((k for k in imgs if os.path.splitext(k)[0] == alt), None)
        if name is None:
            skipped += 1
            continue
        w = float(r.find("size/width").text); h = float(r.find("size/height").text)
        lines = []
        for o in r.findall("object"):
            b = o.find("bndbox")
            x1, y1, x2, y2 = [float(b.find(k).text) for k in ("xmin","ymin","xmax","ymax")]
            lines.append(f"0 {(x1+x2)/2/w:.6f} {(y1+y2)/2/h:.6f} {(x2-x1)/w:.6f} {(y2-y1)/h:.6f}")
        shutil.copy(imgs[name], f"{out}/images/{split}/{name}")
        open(f"{out}/labels/{split}/{os.path.splitext(name)[0]}.txt", "w").write("\n".join(lines))

open(f"{out}/data.yaml", "w").write(
    f"path: {out}\ntrain: images/train\nval: images/val\ntest: images/test\nnames:\n  0: pothole\n")
print("atlanan:", skipped)
```

```python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
model.train(
    data="/content/pothole_yolo/data.yaml",
    epochs=100, imgsz=640, batch=16, patience=20,
    degrees=5, fliplr=0.5, hsv_v=0.4, mosaic=1.0,
    project="/content/drive/MyDrive/pothole/runs", name="exp1",
)
```

```python
metrics = model.val(split="test")
print("mAP50:", metrics.box.map50, "mAP50-95:", metrics.box.map)
```

```python
import glob, pandas as pd, matplotlib.pyplot as plt
from IPython.display import Image, display

# eğitim klasörünü otomatik bul
d = sorted(glob.glob("/content/drive/MyDrive/pothole/runs/**/results.csv", recursive=True))[0].rsplit("/",1)[0]
print("klasör:", d)

# 1) Loss ve metrik grafikleri
df = pd.read_csv(f"{d}/results.csv"); df.columns = df.columns.str.strip()
fig, ax = plt.subplots(2, 3, figsize=(16, 8))
for a, (t, v) in zip(ax[0], [("train/box_loss","val/box_loss"),("train/cls_loss","val/cls_loss"),("train/dfl_loss","val/dfl_loss")]):
    a.plot(df["epoch"], df[t], label="train"); a.plot(df["epoch"], df[v], label="val"); a.set_title(t.split("/")[1]); a.legend()
for a, c in zip(ax[1], ["metrics/precision(B)","metrics/recall(B)","metrics/mAP50-95(B)"]):
    a.plot(df["epoch"], df[c]); a.set_title(c)
plt.tight_layout(); plt.show()

# 2) Test seti grafikleri (PR eğrisi, confusion matrix)
for f in ["PR_curve.png", "confusion_matrix.png"]:
    display(Image(f"/content/runs/detect/val/{f}", width=600))

# 3) Test görüntülerinde tahminler
from ultralytics import YOLO
m = YOLO(f"{d}/weights/best.pt")
r = m.predict("/content/pothole_yolo/images/test", conf=0.25, save=True, max_det=20, verbose=False)
out = sorted(glob.glob(f"{r[0].save_dir}/*.jpg"))[:6]
fig, ax = plt.subplots(2, 3, figsize=(16, 9))
for a, p in zip(ax.ravel(), out):
    a.imshow(plt.imread(p)); a.axis("off")
plt.tight_layout(); plt.show()
```
